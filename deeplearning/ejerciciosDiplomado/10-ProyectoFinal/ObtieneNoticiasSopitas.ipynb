```python
import requests
from bs4 import BeautifulSoup
import operator
# import pandas as pd
import time
```

### Solución completa

```python
sopitasURL='https://www.sopitas.com/'
# dia='27'
# fecha='/2025/03/'+dia+'/'
url = sopitasURL
# url = sopitasURL+fecha

response = requests.get(url)
html_content = response.text
# print(html_content)

soup = BeautifulSoup(html_content, 'html.parser')
sections = soup.find_all('ul', class_='main-menu')
# print("Sections:  ",sections)
categorias=sections[0].find_all('a', class_='')
# print("Categorias:   ",categorias)

urlsCategorias = []
for link in categorias:
    url=link.get("href")
    if "category" in url:
        # urlsCategorias.append(url)
        # for i in range(2,10):
        for i in range(21,30):
            urlsCategorias.append(url+"page/"+str(i))
# print(urlsCategorias)
print("Urls de categorias obtenidas")

noticias=[]
for href in urlsCategorias:
    r = requests.get(href)
    c = r.text
    s = BeautifulSoup(c, 'html.parser')
    noticias.extend(s.find_all('header', class_='entry-header'))
    # print(noticias)
    time.sleep(0.01) #Espera 5 segundos
# print(noticias)
print("Headers obtenidos")

urlsNoticias=[]
for n in noticias:
    urlsNoticias.append(n.find('a').get("href"))
# print(urlsNoticias)
print("Urls de noticias obtenidas: ",len(urlsNoticias))

# with open('./data/Sopitas1_10Pages.txt', 'w') as f:
with open('./data/urlsNoticiasSopitas.txt', 'w') as f:
    for line in urlsNoticias:
        f.write(f"{line}\n")
```

```python
texto=[]
contador=1
for href in urlsNoticias:
    r = requests.get(href)
    c = r.text
    s = BeautifulSoup(c, 'html.parser')
    texto.append(s.find_all('div', class_='entry-content')[0].text)
    # texto.append(s.find_all('div', class_='cabeza')[0].text)
    # texto.append(s.find_all('div', class_='sumarios')[0].text)
    # texto.append(s.find_all('div', class_='text')[0].text)
    # texto.extend(s.find_all('a', class_='cabeza'))
    time.sleep(0.01) #Espera 0.01 segundos
    contador+=1
    # if contador % 10 == 0:
    print("Obtenido:",href,"contador: ",contador)
# print(texto)

# with open('./data/Sopitas1_10Pages.txt', 'w') as f:
with open('./data/Sopitas21_30Pages.txt', 'w') as f:
    for line in texto:
        f.write(f"{line}\n")

print('FIN')
```
