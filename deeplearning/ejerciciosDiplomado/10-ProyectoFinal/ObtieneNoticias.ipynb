```python
import requests
from bs4 import BeautifulSoup
import operator
# import pandas as pd
import time
```

```python
jornadaURL='https://web.jornada.com.mx'
fecha='/2025/03/30/'
```

```python
url = jornadaURL+fecha
response = requests.get(url)
html_content = response.text
# print(html_content)
```

```python
soup = BeautifulSoup(html_content, 'html.parser')
sections = soup.find_all('div', class_='main-sections gui menu')
print(sections[0])
```

```python
categorias=sections[0].find_all('td', class_='')
print(categorias)
```

```python
urlsCategorias = []
for link in categorias:
    url=link.find('a').get("href")
    if fecha in url:
        urlsCategorias.append(jornadaURL+url)
print(urlsCategorias)
```

```python
noticias=[]
for href in urlsCategorias:
    r = requests.get(href)
    c = r.text
    s = BeautifulSoup(c, 'html.parser')
    noticias.extend(s.find_all('a', class_='cabeza'))
    time.sleep(0.5) #Espera 0.01 segundos

print(noticias)
```

```python
urlsNoticias=[]
for n in noticias:
    urlsNoticias.append(jornadaURL+fecha+n.get("href"))
print(urlsNoticias)
```

```python
texto=[]
for href in urlsNoticias:
    r = requests.get(href)
    c = r.text
    s = BeautifulSoup(c, 'html.parser')
    texto.append(s.find_all('div', class_='cabeza')[0].text)
    texto.append(s.find_all('div', class_='sumarios')[0].text)
    texto.append(s.find_all('div', class_='text')[0].text)
    # texto.extend(s.find_all('a', class_='cabeza'))
    time.sleep(0.5) #Espera 0.01 segundos

print(texto)
```

```python
with open('./data/Lajornada'+fecha.replace('/','')+'.txt', 'w') as f:
    for line in texto:
        f.write(f"{line}\n")
```

```python
def cuentaPalabras(texto):
    palabras = texto.split()
    print("Total de palabras:" + str(len(palabras)))
    #Definir arreglo de vocabulario
    vocabulario = {}

    # Analiza cada palabra para igresarla al vocabulario y si ya existe aumenta el contador para conocer su frecuencia
    for palabra in palabras:
        if palabra in vocabulario:
            vocabulario[palabra] += 1
        else:
            vocabulario[palabra] = 1

    return vocabulario
# Mostrar el diccionario resultante
```

```python
f = open('./data/Lajornada'+fecha.replace('/','')+'.txt', "r")
textoArchivo = f.read()
f.close()

vocabulario=cuentaPalabras(textoArchivo)

print("Total de palabras no repetidas:" + str(len(set(vocabulario))))
```

```python
valores_ord = sorted(vocabulario.items(), key=operator.itemgetter(1), reverse=True)
#imprimir la 100 palabras con mayor número de repeticiones
print(valores_ord[0:100])
#imprimir la 100 palabras con menor número de repeticiones
print(valores_ord[-100:])
```

```python
jornadaURL='https://web.jornada.com.mx'
dia='31'

fecha='/2025/03/'+dia+'/'

url = jornadaURL+fecha

response = requests.get(url)

html_content = response.text
# print(html_content)
soup = BeautifulSoup(html_content, 'html.parser')
sections = soup.find_all('div', class_='main-sections gui menu')
# print(sections[0])
categorias=sections[0].find_all('td', class_='')
# print(categorias)

urlsCategorias = []
for link in categorias:
    url=link.find('a').get("href")
    if fecha in url:
        urlsCategorias.append(jornadaURL+url)
print(urlsCategorias)

noticias=[]
for href in urlsCategorias:
    r = requests.get(href)
    c = r.text
    s = BeautifulSoup(c, 'html.parser')
    noticias.extend(s.find_all('a', class_='cabeza'))
    time.sleep(0.1) #Espera 0.01 segundos
print(noticias)

urlsNoticias=[]
for n in noticias:
    urlsNoticias.append(jornadaURL+fecha+n.get("href"))
print(urlsNoticias)

texto=[]
for href in urlsNoticias:
    r = requests.get(href)
    c = r.text
    s = BeautifulSoup(c, 'html.parser')
    texto.append(s.find_all('div', class_='cabeza')[0].text)
    texto.append(s.find_all('div', class_='sumarios')[0].text)
    texto.append(s.find_all('div', class_='text')[0].text)
    # texto.extend(s.find_all('a', class_='cabeza'))
    time.sleep(0.1) #Espera 0.01 segundos
print(texto)

with open('./data/Lajornada'+fecha.replace('/','')+'.txt', 'w') as f:
    for line in texto:
        f.write(f"{line}\n")

print('FIN')
```
