```python
def cuentaPalabras(texto):
    texto=texto.replace(".","")
    texto=texto.replace('"',"")
    texto=texto.replace(',',"")
    texto=texto.replace(':',"")
    texto=texto.replace('(',"")
    texto=texto.replace(')',"")
    texto=texto.replace(';',"")
    texto=texto.replace('“',"")
    texto=texto.replace('”',"")
    texto=texto.replace('¡',"")
    texto=texto.replace('!',"")
    texto=texto.replace('¿',"")
    texto=texto.replace('?',"")
    texto=texto.replace('[',"")
    texto=texto.replace(']',"")
    texto=texto.replace('»',"")
    texto=texto.replace('«',"")
    texto=texto.replace('‘',"")
    # texto=texto.replace('‘',"")
    texto=texto.replace('…',"")
    texto=texto.replace('-',"")
    texto=texto.replace('−',"")
    texto=texto.replace('–',"")



    palabras = texto.split()
    print("Total de palabras:" + str(len(palabras)))
    #Definir arreglo de vocabulario
    vocabulario = {}

    # Analiza cada palabra para igresarla al vocabulario y si ya existe aumenta el contador para conocer su frecuencia
    for palabra in palabras:
        if palabra in vocabulario:
            vocabulario[palabra] += 1
        else:
            vocabulario[palabra] = 1

    return vocabulario
# Mostrar el diccionario resultante
```

```python

# archivoDir="./data/simplebooks/simplebooks-92-raw/train.txt"
# archivoDir="./data/Sopitas1_10Pages.txt"
# archivoDir="./data/Sopitas11_20Pages.txt"
# archivoDir="./data/fakenewsmakerGPT/texto_oraciones_Training.txt"
archivoDir="./data/fakenewsmakerGPT/texto_oraciones_Training_FINAL.txt"

f = open(archivoDir, "r")
textoArchivo = f.read()
f.close()
vocabulario=cuentaPalabras(textoArchivo)
print("Total de palabras no repetidas:" + str(len(set(vocabulario))))


# valores_ord = sorted(vocabulario.items(), key=operator.itemgetter(1), reverse=True)
# print(valores_ord[0:100])

# with open('./data/fakenewsmakerGPT/palabrasSopitas11a20.txt', 'w') as f:
#     for line in valores_ord:
#         f.write(f"{line}\n")
```
