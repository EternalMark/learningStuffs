```python
async def my_async_function(i):
    print(f"Starting task {i}")
    await asyncio.sleep(1)
    print(f"Finishing task {i}")
    return i * 2

async def main():
    tasks = []
    for i in range(5):
        task = asyncio.create_task(my_async_function(i))
        tasks.append(task)
    
    results = await asyncio.gather(*tasks)
    print(f"Results: {results}")
```

# Threads!

```python
import threading
import time
```

```python

def tarea(numero):
    print(f"[Thread {numero}] Iniciando...")
    time.sleep(1)  # Simula trabajo
    print(f"[Thread {numero}] Terminando...")

# Lista para mantener referencia a los threads
threads = []

# Crear 5 threads en un bucle
for i in range(20):
    hilo = threading.Thread(target=tarea, args=(i,))
    hilo.start()
    threads.append(hilo)

# Esperar a que todos los threads terminen
for hilo in threads:
    hilo.join()

print("Todos los threads han terminado.")
```
