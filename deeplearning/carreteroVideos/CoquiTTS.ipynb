```python
carpetaTemporal="temp"+prefijo+"/"
objdir="obj/"

cmd = ['rm','-r',ruta+carpetaTemporal]
subprocess.run(cmd)

os.mkdir(ruta+carpetaTemporal)
os.mkdir(ruta+carpetaTemporal+objdir)

i=1
archivos_a_unir = []
for text in frases:
    nombreOrig=ruta+carpetaTemporal+objdir+prefijo+str(i+3)+"Orig.wav"
    nombreRev=ruta+carpetaTemporal+objdir+prefijo+str(i+3)+"Rev.wav"
    nombre=prefijo+str(i+3)+".wav"
    archivos_a_unir.append(nombre)
    nombre=ruta+carpetaTemporal+nombre
    if i > 1:
        tts.tts_to_file(text=text, file_path=nombreOrig)
        cmd = ['ffmpeg','-i',nombreOrig,'-af','areverse,atrim=start='+('1.3' if i+3 not in pocorecorte else '0.8'),'-c:a','pcm_s16le',nombreRev]
        # cmd = ['ffmpeg','-i',nombreOrig,'-af','areverse,atrim=start='+'1.3','-c:a','pcm_s16le',nombreRev]
        subprocess.run(cmd)
        cmd = ['ffmpeg','-i',nombreRev,'-af','areverse','-c:a','pcm_s16le', nombre]
        subprocess.run(cmd)
    else:
        tts.tts_to_file(text=text, file_path=nombre)
    i+=1

#Generando archivo de 
with open(ruta+carpetaTemporal+'lista_wav.txt', 'w') as f:
    for archivo in archivos_a_unir:
        f.write(f"file '{archivo}'\n")
```
