```python
import numpy as np
# from moviepy.editor import VideoClip

# import numpy as np
from moviepy.video.VideoClip import VideoClip, ImageClip, TextClip
from moviepy.video.compositing.CompositeVideoClip import CompositeVideoClip
print("Version de moviepy ",moviepy.__version__)
```

```python

# Parámetros generales
VIDEO_SIZE = (1280, 720)
FPS = 30
DURATION = 10  # segundos


# Función para crear el fondo hipnótico
def hipnotic_background(t):
    w, h = VIDEO_SIZE
    x = np.linspace(0, 2*np.pi, w)
    y = np.linspace(0, 2*np.pi, h)
    X, Y = np.meshgrid(x, y)

    wave = (np.sin(X * 2 + t) + np.sin(Y * 3 - t*1.5)) * 0.5
    norm_wave = ((wave - wave.min()) / (wave.max() - wave.min()) * 255).astype(np.uint8)

    frame = np.zeros((h, w, 3), dtype=np.uint8)
    frame[..., 0] = norm_wave
    frame[..., 1] = (norm_wave // 2) + 60
    frame[..., 2] = 255 - norm_wave

    return frame
```

```python

# Crear el fondo dinámico (CORREGIDO)
background_clip = VideoClip(make_frame=hipnotic_background, duration=DURATION)

# Crear un texto animado
text_clip = TextClip(
    "¡Bienvenido!", fontsize=70, color='white', font='Arial-Bold'
).set_duration(5).set_position('center')

# Opcional: crear una imagen (si tienes una imagen 'logo.png' en el mismo directorio)
try:
    image_clip = ImageClip('logo.png').set_duration(DURATION).resize(height=100).set_position(('right', 'bottom'))
except OSError:
    image_clip = None  # Si no encuentra la imagen, sigue sin ella

# Componer todos los clips
elements = [background_clip, text_clip]
if image_clip:
    elements.append(image_clip)

final_clip = CompositeVideoClip(elements)

# Exportar el resultado
final_clip.write_videofile("final_video.mp4", fps=FPS)
```

```python
from moviepy import *
import numpy as np

# --- Configuraciones del video ---
ancho = 640
alto = 480
fps = 24
duracion = 5  # Duración en segundos
color_fondo = (255, 105, 180)  # Rosa

# --- Crear el clip de fondo ---
clip_fondo = ColorClip(size=(ancho, alto), color=color_fondo, duration=duracion)

# --- Crear el clip de texto ---
texto = "¡Hola MoviePy!"
clip_texto = TextClip(texto, fontsize=40, color='white', font='Arial')
clip_texto = clip_texto.set_pos('center').set_duration(duracion)

# --- Efecto de aparición y desaparición del texto ---
def efecto_fadeinout(t):
    if t < 1:
        return t  # Aparece gradualmente en el primer segundo
    elif t > duracion - 1:
        return duracion - t  # Desaparece gradualmente en el último segundo
    else:
        return 1  # Completamente visible en el medio

clip_texto = clip_texto.set_opacity(clip_texto.fl_time(efecto_fadeinout))

# --- Componer los clips ---
video_final = CompositeVideoClip([clip_fondo, clip_texto], size=(ancho, alto))

# --- Escribir el archivo de video ---
nombre_archivo = "video_desde_cero.mp4"
video_final.write_videofile(nombre_archivo, fps=fps)

print(f"¡Video '{nombre_archivo}' creado exitosamente!")
```

# Pegando palabras en un video

```python
from moviepy.editor import VideoFileClip, TextClip, CompositeVideoClip
from moviepy.video.fx.fadein import fadein
from moviepy.video.fx.fadeout import fadeout
from moviepy.video.tools.subtitles import SubtitlesClip
```

```python
# Cargar el video base
video = VideoFileClip("tu_video.mp4")

# Texto a mostrar
texto = "Hola, viajero del tiempo..."

# Generar una lista de clips con cada letra apareciendo una a una
duracion_total = 3  # segundos que tarda en escribirse todo
duracion_letra = duracion_total / len(texto)
clips_letras = []

for i, letra in enumerate(texto):
    if letra == " ":
        continue  # No mostramos espacios
    t_inicio = i * duracion_letra
    clip = (TextClip(letra, fontsize=70, color='white', font='Arial-Bold')
            .set_start(t_inicio)
            .set_duration(1.5)  # duración visible de la letra
            .set_pos(('center', 'bottom'))
            .fadein(0.3)
            .fadeout(0.3))
    clips_letras.append(clip)

# Combinar todos los clips de letras sobre el video
video_con_texto = CompositeVideoClip([video] + clips_letras)

# Cortar video si es más largo que el texto
final = video_con_texto.set_duration(video.duration)

# Exportar resultado
final.write_videofile("video_con_texto.mp4", fps=24)
```
